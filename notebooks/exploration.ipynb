```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Pointing exactly to your 'data.csv' file inside your folder structure
df = pd.read_csv('../data/data.csv')

print("Dataset Shape:", df.shape)
print("Feature Names:", df.columns.tolist())
display(df.head())
```

```python
display(df.describe())
print("Class Distribution:")
print(df['diagnosis'].value_counts())
```

```python
# Sum up all missing values per column
missing_values = df.isnull().sum()
print("Missing values per feature:\n", missing_values)
```

```python
import matplotlib.pyplot as plt

# 1. Target Distribution Bar Chart (Malignant vs. Benign)
plt.figure(figsize=(6,4))
df['diagnosis'].value_counts().plot(kind='bar', color=['#4C72B0', '#C44E52'], edgecolor='black')
plt.title('Distribution of Breast Cancer Diagnoses')
plt.xlabel('Diagnosis (B = Benign, M = Malignant)')
plt.ylabel('Patient Count')
plt.xticks(rotation=0) # Keeps the B and M text upright
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 2. Key Feature Distribution (Cellular Radius Mean)
plt.figure(figsize=(6,4))
plt.hist(df['radius_mean'], bins=25, color='#55A868', edgecolor='black', alpha=0.8)
plt.title('Distribution of Cellular Radius Mean')
plt.xlabel('Radius Mean Value')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

df_numeric = df.copy()


columns_to_drop = ['Unnamed: 0', 'id']
for col in columns_to_drop:
    if col in df_numeric.columns:
        df_numeric = df_numeric.drop(columns=[col])


df_numeric['diagnosis'] = df_numeric['diagnosis'].map({'M': 1, 'B': 0})

# 4. Filter out any remaining non-numeric columns to prevent math errors
df_numeric = df_numeric.select_dtypes(include=['number'])

# 5. Compute Pearson correlation values
correlation_matrix = df_numeric.corr()

top_correlated_features = correlation_matrix['diagnosis'].sort_values(ascending=False).index[:11]


plt.figure(figsize=(12, 10))
sns.heatmap(df_numeric[top_correlated_features].corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Top Feature Correlations with Malignancy Diagnosis')
plt.show()


print("Correlation with the Diagnosis Variable:")
print(correlation_matrix['diagnosis'].sort_values(ascending=False))
```
